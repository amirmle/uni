```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def split_random(x, y, test_frac=0.2, seed=0):
    rng = np.random.RandomState(seed)
    n = x.shape[0]
    idx = np.arange(n)
    rng.shuffle(idx)
    n_test = int(np.ceil(test_frac * n))
    test_idx = idx[:n_test]
    train_idx = idx[n_test:]
    return x[train_idx], y[train_idx], x[test_idx], y[test_idx]
    
```

```python
def mse_error(y_true, y_pred):
    return np.mean((y_true - y_pred)**2)
```

```python
def ols_solution(Phi, t):
    A = Phi.T.dot(Phi)
    b = Phi.T.dot(t)
    w = np.linalg.solve(A, b)
    return w
```

```python
def predict(Phi, w):
    return Phi.dot(w).flatten()
```

```python
def design_matrix_gaussian(X, centers, s):
    N = X.shape[0]
    M = centers.shape[0]
    Phi = np.ones((N, 1 + M))
    for j in range(M):
        diff = X - centers[j:j+1, :]
        norms2 = np.sum(diff**2, axis=1)
        Phi[:, 1+j] = np.exp(-norms2 / (2*(s**2)))
    return Phi
```
