```python
import numpy as np
```

```python
data = [
    [293,1.6,851],
    [230,15.5,816],
    [172,22.0,1058],
    [91,43.0,1201],
    [113,33.0,1357],
    [125,40.0,1115],
]
data = np.array(data)
y = data[:,0]
x = data[:,1:]
x = np.column_stack((np.ones(x.shape[0]),x))
x
```

```python
def multivariate_regression (x,y):
    a = x.T@x
    a_inv = np.linalg.inv(a)
    b = x.T@y
    return a_inv@b
```

```python
w = multivariate_regression(x,y)
w
```

```python
x_test = [1,25,1000]
y_test = w.T @ x_test
y_test
```

```python
x_interactive = x[:,1]*x[:,2]
x_new = np.column_stack((x,x_interactive))
x_new
```

```python
w_new = multivariate_regression(x_new,y)
w_new
```

```python
x_test_2 = [1,25,1000,2500]
y_test_2 = w_new.T @ x_test_2
y_test_2
```
