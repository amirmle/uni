```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.io.wavfile as wav
```

```python
sample_rate, data = wav.read('harvard.wav')
print('sample rate:', sample_rate)
print('data shape:', data.shape)
```

```python
plt.clf()

channel0 = data[:,0]
channel1 = data[:,1]
plt.plot( channel0)
plt.plot( channel1)
plt.xlabel('sample rate')
plt.show()
```

```python
plt.clf()

N , no_channels = data.shape
time = np.arange(N)/sample_rate
print(time)
plt.plot(time, channel0)
plt.plot(time, channel1)
plt.xlabel('time [s]')
plt.show()
```

## Analyse the audio signal
 سیگنال صوتی نواسانات دامنه بر حسب زمان است
 در نمودار، نواحی با نوسان شدید نشون دهنده بخش هایی که در آن صدا وجود داره است
 نواحی صاف یا نزدیک به صفر مربوط به سکوت بین کلمات یا جملات است
 اگر به صوت گوش بدیم اونجا هایی که صدا بلند تر یا شدیدتر میشه تو نمودار به صورت قله نمایش داده شده

## Plot both channels using a single plt.plot function
 با ابزار زوم میشه رو بخشی از سیگنال زوم کرد
 وقتی زوم میکنیم میشه دید سیگنال از نقاط نوسانی نسبتا منظمی تشکیل شده اما خب کاملا سینوسی نیست

```python
plt.clf()
plt.plot(data)
plt.show()
```

```python
plt.clf()

scale = 2**np.linspace(-2,4, N)
scale = scale.reshape(N, 1)

data_new = np.int16(scale * data)
channel0_new = data_new[:,0]
channel1_new = data_new[:,1]

plt.plot(time, channel0_new)
plt.plot(time, channel1_new)

plt.xlabel('time [s]')
plt.show()
```

 در تغییر بالا دامنه نوسان تغییر کرده و در ضریبی که در مثال های قبل بود ضرب شده
 اول سیگنال ، سیگنال های در ضریبی کوچکتر از یک ضرب شده و کاهش شدت صدا داریم و در ادامه شدت صدا افزایش پیدا میکنه
