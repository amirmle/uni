```python
import numpy as np
import scipy.io.wavfile
import matplotlib.pyplot as plt
```

```python
file_name = 'harvard.wav'
sampling_rate, data = scipy.io.wavfile.read(file_name)
print('samping rate:', sampling_rate) # frequency (sample per second)
print('data type:', data.dtype)
print('data shape:', data.shape)
N, no_channels = data.shape # signal length and no. of channels
print('signal length:', N)
channel0 = data[:,0]
channel1 = data[:,1]
```

```python
plt.clf()
plt.plot(channel0)
plt.plot(channel1)
plt.title('Original signal')
plt.show()
```

```python
plt.clf()
new_data = np.flip(data, axis=0)
# reversed_channel0 = channel0[::-1]
# reversed_channel1 = channel1[::-1]
reversed_channel0 = new_data[:,0]
reversed_channel1 = new_data[:,1]
plt.plot(reversed_channel0)
plt.plot(reversed_channel1)
plt.title('reversed signal')
plt.show()
```

```python
output_file_name = f'reversed_{file_name}'
scipy.io.wavfile.write(output_file_name, sampling_rate, new_data)
print('Done')
```
