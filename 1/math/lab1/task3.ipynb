```python
import numpy as np
import scipy.io.wavfile
import matplotlib.pyplot as plt
```

```python
file_name = 'harvard.wav'
sampling_rate, data = scipy.io.wavfile.read(file_name)
print('samping rate:', sampling_rate) # frequency (sample per second)
print('data type:', data.dtype)
print('data shape:', data.shape)
N, no_channels = data.shape # signal length and no. of channels
print('signal length:', N)
channel0 = data[:,0]
channel1 = data[:,1]
```

```python
sampling_rate_1 = sampling_rate//2
sampling_rate_2 = sampling_rate*2
print('sampling rate // 2:', sampling_rate_1)
print('slower sampling rate * 2:', sampling_rate_2)

output_1_file_name = f'sampling_rate%2{file_name}'
scipy.io.wavfile.write(output_1_file_name, sampling_rate_1, data)
print(f'{output_1_file_name}Done')

output_2_file_name = f'sampling_ratex2{file_name}'
scipy.io.wavfile.write(output_2_file_name, sampling_rate_2, data)
print(f'{output_2_file_name}Done')
```
