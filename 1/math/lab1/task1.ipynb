```python
import numpy as np
```

```python
A = np.random.rand(200,10)
A
```

  قطعه کد زیر  میانگین هر ستون را بدست میاورد و یک آرایه با ده مقدار خروجی میدهد

```python
mu = np.zeros(A.shape[1])


for i in range(A.shape[0]):
    mu += A[i]

mu /= A.shape[0]
mu
```

 و در کد زیر اختلاف هر خانه با میانگین ستون متناظر در آن را بدست میاورد

```python
B = np.zeros_like(A)
for i in range(A.shape[0]):
    B[i] = A[i] - mu
B
```

 equivalent program without loops just in a single line of code

```python
diff = A - A.mean(axis=0)

diff
```
