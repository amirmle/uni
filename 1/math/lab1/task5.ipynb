```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = plt.imread('image.png')
np_img = np.array(img)
plt.close()
plt.imshow(np_img)
```

```python
flipped = np_img[::-1]

plt.clf()
plt.imshow(flipped)
```

```python
union = np.vstack((np_img, flipped))

plt.clf() 
plt.imshow(union)
```
