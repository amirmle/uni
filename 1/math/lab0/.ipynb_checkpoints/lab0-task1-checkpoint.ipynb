```python
def mul(a : list,b : list):
    a_x_len = len(a)
    a_y_len = len(a[0])
    b_x_len = len(b)
    b_y_len = len(b[0])
    if a_y_len != b_x_len :
        return []
    
    c = [[0 for _ in range(b_y_len)] for _ in range(a_x_len)]

            
    
    for i in range(a_x_len):
        for j in range(b_y_len):
            for k in range(a_y_len):
                c[i][j] += a[i][k] * b[k][j]
    
    return c
    
```

```python

A = [[1, 0, 0], [0, 0, 3], [0, 2, 0]]
B = [[1, 1], [0, .5], [2, 1/3.0]]
C = [[ 1, 0, 0 ], [ 0, 0, 0.5], [ 0, 1/3.0, 0]]


print(mul(A,B))
print(mul(B,A))
print(mul(A,C))
```
